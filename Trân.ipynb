```python
import pandas as pd

geography_raw = pd.read_csv("geography.csv")
customers_raw = pd.read_csv("customers.csv")
product_epd_raw = pd.read_json("epd.json")
promotion_eprom_raw = pd.read_json("eprom.json")
shipments_realistic_raw = pd.read_csv("shipments_realistic.csv")
tf_raw = pd.read_json("tf.json")
web_traffic_raw = pd.read_csv("web_traffic.csv")
```

```python

# GEOGRAPHY

geography = geography_raw[
    [
        "zip",
        "city",
        "region",
        "district"
    ]
].copy()

# Chuẩn hóa text
for col in [
    "city",
    "region",
    "district"
]:
    geography[col] = (
        geography[col]
        .astype("string")
        .str.strip()
    )

# Chuẩn hóa ZIP
geography["zip"] = pd.to_numeric(
    geography["zip"],
    errors="coerce"
).astype("Int64")

# Xóa ZIP bị thiếu
geography = geography.dropna(
    subset=["zip"]
)

# Xóa ZIP trùng
geography = geography.drop_duplicates(
    subset=["zip"]
).reset_index(drop=True)

# Kiểm tra
print("Geography:")
print(geography.info())
print("\nNull:")
print(geography.isnull().sum())
print("\nDuplicate zip:",
      geography["zip"].duplicated().sum())

geography.head()
```

```python

# CUSTOMERS


customers = customers_raw[
    [
        "customer_id",
        "zip",
        "city",
        "signup_date",
        "gender",
        "age_group",
        "acquisition_channel"
    ]
].copy()

# Chuẩn hóa dữ liệu text
for col in [
    "city",
    "gender",
    "age_group",
    "acquisition_channel"
]:
    customers[col] = (
        customers[col]
        .astype("string")
        .str.strip()
    )

# Chuẩn hóa ID và ZIP
customers["customer_id"] = pd.to_numeric(
    customers["customer_id"],
    errors="coerce"
).astype("Int64")

customers["zip"] = pd.to_numeric(
    customers["zip"],
    errors="coerce"
).astype("Int64")

# Chuẩn hóa ngày
customers["signup_date"] = pd.to_datetime(
    customers["signup_date"],
    errors="coerce"
)

# Xóa customer không có ID
customers = customers.dropna(
    subset=["customer_id"]
)

# Xóa customer bị trùng ID
customers = customers.drop_duplicates(
    subset=["customer_id"]
).reset_index(drop=True)

# Kiểm tra
print("Customers:")
print(customers.info())
print("\nNull:")
print(customers.isnull().sum())
print("\nDuplicate customer_id:",
      customers["customer_id"].duplicated().sum())

customers.head()
```

```python

# EPD - PRODUCT


product = product_epd_raw[
    [
        "product_id",
        "product_name",
        "category",
        "segment",
        "size",
        "color",
        "price",
        "cogs"
    ]
].copy()

# Chuẩn hóa text
for col in [
    "product_name",
    "category",
    "segment",
    "size",
    "color"
]:
    product[col] = (
        product[col]
        .astype("string")
        .str.strip()
    )

# Chuẩn hóa Product ID
product["product_id"] = pd.to_numeric(
    product["product_id"],
    errors="coerce"
).astype("Int64")

# Chuẩn hóa giá
product["price"] = pd.to_numeric(
    product["price"],
    errors="coerce"
)

# Chuẩn hóa giá vốn
product["cogs"] = pd.to_numeric(
    product["cogs"],
    errors="coerce"
)

# Xóa Product không có ID
product = product.dropna(
    subset=["product_id"]
)

# Xóa Product ID trùng
product = product.drop_duplicates(
    subset=["product_id"]
).reset_index(drop=True)

# Kiểm tra
print("Product:")
print(product.info())
print("\nNull:")
print(product.isnull().sum())
print("\nDuplicate product_id:",
      product["product_id"].duplicated().sum())

product.head()
```

```python

# EPROM - PROMOTION


promotion = promotion_eprom_raw[
    [
        "promo_id",
        "promo_name",
        "promo_type",
        "discount_value",
        "start_date",
        "end_date",
        "applicable_category",
        "promo_channel",
        "stackable_flag",
        "min_order_value"
    ]
].copy()

# Chuẩn hóa text
for col in [
    "promo_id",
    "promo_name",
    "promo_type",
    "applicable_category",
    "promo_channel"
]:
    promotion[col] = (
        promotion[col]
        .astype("string")
        .str.strip()
    )

# Chuẩn hóa số
promotion["discount_value"] = pd.to_numeric(
    promotion["discount_value"],
    errors="coerce"
)

promotion["stackable_flag"] = pd.to_numeric(
    promotion["stackable_flag"],
    errors="coerce"
).astype("Int64")

promotion["min_order_value"] = pd.to_numeric(
    promotion["min_order_value"],
    errors="coerce"
)

# Chuẩn hóa ngày
promotion["start_date"] = pd.to_datetime(
    promotion["start_date"],
    errors="coerce"
)

promotion["end_date"] = pd.to_datetime(
    promotion["end_date"],
    errors="coerce"
)

# Xóa Promotion không có ID
promotion = promotion.dropna(
    subset=["promo_id"]
)

# Xóa ID trùng
promotion = promotion.drop_duplicates(
    subset=["promo_id"]
).reset_index(drop=True)

# Kiểm tra
print("Promotion:")
print(promotion.info())
print("\nNull:")
print(promotion.isnull().sum())
print("\nDuplicate promo_id:",
      promotion["promo_id"].duplicated().sum())

promotion.head()
```

```python

# SHIPMENTS RAW CLEANING


ship_temp = shipments_realistic_raw.copy()

# Chuẩn hóa tên cột
ship_temp.columns = (
    ship_temp.columns
    .str.strip()
)

# Chuẩn hóa text
text_columns = [
    "shipper_id",
    "shipper_company",
    "shipper_vehicle",
    "working_shift",
    "shipper_name",
    "shipper_phone",
    "shipper_gender",
    "shipper_marital_status",
    "shipper_education",
    "city",
    "region",
    "district"
]

for col in text_columns:
    ship_temp[col] = (
        ship_temp[col]
        .astype("string")
        .str.strip()
    )

# Chuẩn hóa ID
ship_temp["order_id"] = pd.to_numeric(
    ship_temp["order_id"],
    errors="coerce"
).astype("Int64")

# Chuẩn hóa số
numeric_columns = [
    "shipping_fee",
    "shipper_experience_years",
    "shipper_rating",
    "delivery_success_rate",
    "average_delivery_time",
    "shipper_age"
]

for col in numeric_columns:
    ship_temp[col] = pd.to_numeric(
        ship_temp[col],
        errors="coerce"
    )

# Chuẩn hóa ngày
ship_temp["ship_date"] = pd.to_datetime(
    ship_temp["ship_date"],
    errors="coerce"
)

ship_temp["delivery_date"] = pd.to_datetime(
    ship_temp["delivery_date"],
    errors="coerce"
)

ship_temp["join_date"] = pd.to_datetime(
    ship_temp["join_date"],
    errors="coerce"
)
```

```python

# SHIPPERS


shippers = ship_temp[
    [
        "shipper_id",
        "shipper_company",
        "shipper_vehicle",
        "shipper_experience_years",
        "shipper_rating",
        "delivery_success_rate",
        "average_delivery_time",
        "working_shift",
        "join_date",
        "shipper_name",
        "shipper_phone",
        "shipper_gender",
        "shipper_age",
        "shipper_marital_status",
        "shipper_education",
        "city",
        "region",
        "district"
    ]
].copy()

# Xóa Shipper ID thiếu
shippers = shippers.dropna(
    subset=["shipper_id"]
)

# Xóa Shipper trùng
shippers = shippers.drop_duplicates(
    subset=["shipper_id"]
).reset_index(drop=True)

shippers.head()
```

```python

# ADD ZIP TO SHIPPERS


geo_map = geography[
    [
        "zip",
        "city",
        "region",
        "district"
    ]
].copy()

# Tạo bản sao để chuẩn hóa cho việc merge
geo_map["city"] = (
    geo_map["city"]
    .astype("string")
    .str.strip()
    .str.lower()
)

geo_map["region"] = (
    geo_map["region"]
    .astype("string")
    .str.strip()
    .str.lower()
)

geo_map["district"] = (
    geo_map["district"]
    .astype("string")
    .str.strip()
    .str.lower()
)

shippers["city"] = (
    shippers["city"]
    .astype("string")
    .str.strip()
    .str.lower()
)

shippers["region"] = (
    shippers["region"]
    .astype("string")
    .str.strip()
    .str.lower()
)

shippers["district"] = (
    shippers["district"]
    .astype("string")
    .str.strip()
    .str.lower()
)

geo_map = geo_map.drop_duplicates(
    subset=["city", "region", "district"]
)

shippers = shippers.merge(
    geo_map,
    on=["city", "region", "district"],
    how="left"
)

shippers.head()
```

```python
print(
    "Số Shipper không tìm được ZIP:",
    shippers["zip"].isna().sum()
)
```

```python

# SHIPMENTS


shipments = ship_temp[
    [
        "order_id",
        "shipper_id",
        "ship_date",
        "delivery_date",
        "shipping_fee"
    ]
].copy()

# Xóa Order không có ID
shipments = shipments.dropna(
    subset=["order_id"]
)

# Xóa Order bị trùng
shipments = shipments.drop_duplicates(
    subset=["order_id"]
).reset_index(drop=True)

shipments.head()
```

```python

# CHECK SHIPPER ID


invalid_shipper = shipments[
    ~shipments["shipper_id"].isin(
        shippers["shipper_id"]
    )
]

print(
    "Số Shipment có shipper_id không tồn tại:",
    len(invalid_shipper)
)
```

```python
# TF

tf = tf_raw[
    [
        "date",
        "sessions",
        "unique_visitors",
        "page_views",
        "bounce_rate",
        "avg_session_duration_sec",
        "traffic_source"
    ]
].copy()

# Chuẩn hóa ngày
tf["date"] = pd.to_datetime(
    tf["date"],
    errors="coerce"
)

# Chuẩn hóa số
numeric_columns = [
    "sessions",
    "unique_visitors",
    "page_views",
    "bounce_rate",
    "avg_session_duration_sec"
]

for col in numeric_columns:
    tf[col] = pd.to_numeric(
        tf[col],
        errors="coerce"
    )

# Chuẩn hóa traffic source
tf["traffic_source"] = (
    tf["traffic_source"]
    .astype("string")
    .str.strip()
)

# Xóa ngày thiếu
tf = tf.dropna(
    subset=["date"]
).reset_index(drop=True)

tf.head()
```

```python
tf["traffic_source"] = (
    tf["traffic_source"]
    .replace("", pd.NA)
)
```

```python
# WEB TRAFFIC


web_traffic = web_traffic_raw[
    [
        "date",
        "sessions",
        "unique_visitors",
        "page_views",
        "bounce_rate",
        "avg_session_duration_sec",
        "traffic_source"
    ]
].copy()

# Chuẩn hóa ngày
web_traffic["date"] = pd.to_datetime(
    web_traffic["date"],
    errors="coerce"
)

# Chuẩn hóa số
numeric_columns = [
    "sessions",
    "unique_visitors",
    "page_views",
    "bounce_rate",
    "avg_session_duration_sec"
]

for col in numeric_columns:
    web_traffic[col] = pd.to_numeric(
        web_traffic[col],
        errors="coerce"
    )

# Chuẩn hóa text
web_traffic["traffic_source"] = (
    web_traffic["traffic_source"]
    .astype("string")
    .str.strip()
)

# Chuyển chuỗi rỗng thành NA
web_traffic["traffic_source"] = (
    web_traffic["traffic_source"]
    .replace("", pd.NA)
)

# Xóa ngày thiếu
web_traffic = web_traffic.dropna(
    subset=["date"]
).reset_index(drop=True)

web_traffic.head()
```
